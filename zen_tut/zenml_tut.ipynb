```python
from zenml import step, pipeline

# STEP 1 — load data
@step
def load_data() -> list:
    return [1, 2, 3, 4, 5]

# STEP 2 — process data
@step
def multiply(values: list) -> list:
    return [v * 2 for v in values]

# STEP 3 — final result
@step
def trainer(values: list) -> int:
    return sum(values)

# PIPELINE
@pipeline
def my_first_pipeline():
    data = load_data()
    processed = multiply(data)
    trainer(processed)
```

```python
my_first_pipeline()
```
