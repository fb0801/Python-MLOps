```python
from zenml import step, pipeline

# Step 1
@step
def load_data() -> list:
    return [1, 2, 3, 4]

# Step 2
@step
def multiply(data: list) -> list:
    return [x * 2 for x in data]

# Step 3
@step
def trainer(data: list) -> int:
    return sum(data)

# Pipeline
@pipeline
def my_pipeline():
    data = load_data()
    processed = multiply(data)
    trainer(processed)
```

```python
my_pipeline()
```
